### 2. Pembersihan Data
---

```python
missing_pct = data.isna().sum() / len(data)
cols_to_drop = missing_pct[missing_pct > 0.5]
cols_to_drop = cols_to_drop.index.tolist()

data = data.drop(columns=cols_to_drop, axis=1)
data.isna().sum()

print(cols_to_drop)
missing_pct
```

```python
#Kita imputasi yang tipe datanya angka dulu
data.info()
```

```python
num_imputer = SimpleImputer(strategy="median")
obj_imputer = SimpleImputer(strategy="most_frequent")

cols_num = data.select_dtypes(include=["number"]).columns
cols_num_missing = cols_num[data[cols_num].isna().sum() > 0].to_list()

data[cols_num_missing] = num_imputer.fit_transform(data[cols_num_missing])

cols_obj = data.select_dtypes(include=["object"]).columns
cols_obj_missing = cols_obj[data[cols_obj].isna().sum() > 0].to_list()

data[cols_obj_missing] = obj_imputer.fit_transform(data[cols_obj_missing])

data.isna().sum()
```

### 3. Deteksi dan Penanganan Outlier
---

```python
features = cols_num.drop("Id")  # kita drop fitur Id
n_feature = len(features)
n_cols = 3
n_rows = int(np.ceil(n_feature / 3))

fig, axes = plt.subplots(n_rows, n_cols, figsize=(7 * n_cols, 5 * n_rows))
axes = axes.flatten()

for col, ax in zip(features, axes):
  sns.boxplot(y=col, data=data, ax=ax)
  ax.set_title(f"BoxPlot {col}")
  ax.set_ylabel(col)

for ax in axes[n_feature:]:
  fig.delaxes(ax)

plt.tight_layout()
plt.show()
```

```python
data = data_copy.copy()

encoder = OneHotEncoder(sparse_output=False, handle_unknown="ignore")

encoded_matrix_obj = encoder.fit_transform(data[cols_obj])
encoded_cols_obj = encoder.get_feature_names_out(cols_obj)
encoded_df_obj = pd.DataFrame(encoded_matrix_obj, columns=encoded_cols_obj, index=data.index)

data = data.drop(columns=cols_obj, axis=1)
data = pd.concat([data, encoded_df_obj], axis=1)
```

```python
scaler = StandardScaler()

x_data, y_data = data.drop(columns=["SalePrice"], axis=1), data["SalePrice"]
x_columns = x_data.columns
x_data = scaler.fit_transform(x_data)
```

### 5. Reduksi Dimensi
---

```python
x_data = pd.DataFrame(x_data, columns=x_columns)
corr_matrix = x_data.corr().abs()
upper_triangle = corr_matrix.where(np.triu(np.ones(corr_matrix.shape), k=1).astype(bool))

threshold = 0.85
to_drop = [column for column in upper_triangle.columns if any(upper_triangle[column] > threshold)]

print(f"Fitur yang sangat berkorelasi (> {threshold}) dan akan dibuang:")
print(to_drop)

x_data_reduced = x_data.drop(columns=to_drop)
```

```python
pca = PCA(n_components=0.9)
x_pca = pca.fit_transform(x_data_reduced)
```
